# DDA Line Drawing

```python
import matplotlib.pyplot as plt 

def dda(x1, y1, x2, y2):
    dx = x2 - x1 
    dy = y2 - y1 

    step = int(max(abs(dx), abs(dy)))

    xs = dx / step
    ys = dy / step

    x, y = x1, y1

    x_nw = []
    y_nw = []

    for i in range(step + 1):
        x_nw.append(round(x))
        y_nw.append(round(y))
        x = x + xs 
        y = y + ys 

    return x_nw, y_nw

x1 = int(input('Enter x1: '))
y1 = int(input('Enter y1: '))
x2 = int(input('Enter x2: '))
y2 = int(input('Enter y2: '))

x, y = dda(x1, y1, x2, y2) 

plt.plot(x, y, 'r-')
plt.grid()
plt.show()
```

# Bresenham Line Drawing

```python
import matplotlib.pyplot as plt

def bresenham(x1, y1, x2, y2): 
    dx = x2 - x1
    dy = y2 - y1

    p = 2 * dy - dx 

    x, y = x1, y1 

    x_nw = [x1]
    y_nw = [y1]
    
    while x < x2:
        if p < 0: 
            x += 1
            p = p + 2 * dy 
            x_nw.append(x) 
            y_nw.append(y)

        else:
            x += 1
            y += 1
            p = p + 2 * dy - 2 * dx 
            x_nw.append(x)
            y_nw.append(y)
    return x_nw, y_nw

x1 = int(input('Enter x1: '))
y1 = int(input('Enter y1: '))
x2 = int(input('Enter x2: '))
y2 = int(input('Enter y2: '))

x, y = bresenham(x1, y1, x2, y2)

for xi, yi in zip(x, y):
    print(f'{(xi, yi)}')

plt.plot(x, y, 'r', marker='o')
plt.grid() 
plt.show()
```

# Bresenham Circle Drawing

```python
import matplotlib.pyplot as plt 

def brecircle(xc, yc):
    r = int(input('Enter r : '))

    x = 0
    y = r 
    d = 3 - 2 * r

    x_nw = []
    y_nw = []
    
    while x <= y:
        points = [
            (x + xc, y + yc),
            (y + xc, x + yc), 
            (-x + xc, y + yc), 
            (-y + xc, x + yc), 
            (-x + xc, -y + yc), 
            (-y + xc, -x + yc),
            (x + xc, -y + yc),
            (y + xc, -x + yc)
        ]

        for xi, yi in points:
            x_nw.append(xi)
            y_nw.append(yi)

        if d < 0:
            d = d + 4 * x + 6
        else:
            d = d + 4 * (x - y) + 10
            y -= 1

        x += 1

    return x_nw, y_nw

# input center
xc = int(input('Enter xc: '))
yc = int(input('Enter yc: '))

x, y = brecircle(xc, yc)

for xi, yi in zip(x, y):
    print((xi, yi))

plt.scatter(x, y)
plt.gca().set_aspect('equal')
plt.grid() 
plt.show()
```

# Mid-Point Circle Drawing

```python
def mid(xc, yc, r):
    x = 0
    y = r 
    d = 1 - r 
    x_nw = []
    y_nw = []
    while x <= y:
        points = [
            (x + xc, y + yc),
            (y + xc, x + yc),
            (-x + xc, y + yc),
            (-y + xc, x + yc),
            (-x + xc, -y + yc),
            (-y + xc, -x + yc),
            (x + xc, -y + yc),
            (y + xc, -x + yc)
        ]
        for xi, yi in points:
            x_nw.append(xi) 
            y_nw.append(yi) 

        if d < 0:
            x += 1
            d = d + 2 * x + 1

        else:
            x += 1
            y -= 1
            d = d + 2 * x - 2 * y + 1

    return x_nw, y_nw


    # input center
xc = int(input('Enter xc: '))
yc = int(input('Enter yc: '))
r = int(input('Enter r : '))
x, y = mid(xc, yc, r)

for xi, yi in zip(x, y):
    print((xi, yi))

plt.scatter(x, y)
plt.gca().set_aspect('equal')
plt.grid() 
plt.show()
```

# 2D Rotation

```python
import matplotlib.pyplot as plt
import math
x = [2, 7, 4, 2]
y = [2, 2, 6, 2]

angle = int(input('Enter Angle : '))
radi = math.radians(angle)

xc = sum(x) / len(x) 
yc = sum(y) / len(y)

# Correct translation
x_nw = [(xi - xc) * math.cos(radi) - (yi -yc) * math.sin(radi) + xc for xi, yi in zip(x, y)]
y_nw = [(xi - xc) * math.sin(radi) + (yi -yc) * math.cos(radi) + yc for xi, yi in zip(x, y)]

# Plot
plt.fill(x, y, 'g', alpha=0.2, label='Original')
plt.fill(x_nw, y_nw, 'r', alpha=0.2, label='Translated')

plt.plot(x, y, 'black')
plt.plot(x_nw, y_nw, 'black')

plt.legend()
plt.grid()
plt.show()
```

# Line Clipping

```python
import matplotlib.pyplot as plt

xw = [2, 4, 4, 2, 2]
yw = [2, 2, 4, 4, 2]

x_min, x_max = 2, 4
y_min, y_max = 2, 4

INSIDE = 0
LEFT = 1 
RIGHT = 2 
BOTTOM = 4
TOP = 8

def compute_code(x, y):
    code = INSIDE

    if x < x_min:
        code |= LEFT
    if x > x_max:
        code |= RIGHT
    if y < y_min:
        code |= BOTTOM
    if y > y_max:
        code |= TOP

    return code

def cohen(x1, y1, x2, y2):
    code1 = compute_code(x1, y1)
    code2 = compute_code(x2, y2)

    while True:
        if code1 == 0 and code2 == 0:
            return x1, y1, x2, y2   # accept

        elif (code1 & code2) != 0:
            return None             # reject

        else:
            if code1 != 0:
                outside = code1
            else:
                outside = code2

            if outside & TOP:
                x = x1 + (x2 - x1) * (y_max - y1) / (y2 - y1)
                y = y_max

            elif outside & BOTTOM:
                x = x1 + (x2 - x1) * (y_min - y1) / (y2 - y1)
                y = y_min

            elif outside & RIGHT:
                y = y1 + (y2 - y1) * (x_max - x1) / (x2 - x1)
                x = x_max

            elif outside & LEFT:
                y = y1 + (y2 - y1) * (x_min - x1) / (x2 - x1)
                x = x_min

            if outside == code1:
                x1, y1 = x, y
                code1 = compute_code(x1, y1)
            else:
                x2, y2 = x, y
                code2 = compute_code(x2, y2)

# ---- INPUT ----
x1 = float(input("Enter x1: "))
y1 = float(input("Enter y1: "))
x2 = float(input("Enter x2: "))
y2 = float(input("Enter y2: "))

result = cohen(x1, y1, x2, y2)

# ---- PLOT ----
plt.fill(xw, yw, 'r', alpha=0.2)
plt.plot(xw, yw, 'black')

plt.plot([x1, x2], [y1, y2], 'r--', label='Original')

if result:
    x1c, y1c, x2c, y2c = result
    plt.plot([x1c, x2c], [y1c, y2c], 'g', linewidth=3, label='Clipped')

plt.legend()
plt.title("Cohen-Sutherland Line Clipping")
plt.grid()
plt.show()
```

# Curve Clipping

```python
import matplotlib.pyplot as plt 
import numpy as np 
xw = [2, 10, 10, 2, 2]
yw = [2, 2, 10, 10, 2] 

xmin = min(xw)
xmax = max(xw)
ymin = min(yw) 
ymax = max(yw) 

x = np.linspace(-10, 10, 300)
y = x ** 2 / 5 

inside = [] 
outside = []

for xi, yi in zip(x, y):
    if xmin <= xi <= xmax and ymin <= yi <= ymax:
        inside.append((xi, yi))
    else:
        outside.append((xi, yi))

plt.fill(xw, yw, 'g', alpha=0.2)
plt.plot(xw, yw, 'black')

if inside: 
    for xi, yi in inside:
        plt.scatter(xi, yi, color='green', s=2)
if outside: 
    for xi, yi in outside:
        plt.scatter(xi, yi, color='red', s=2)   

plt.grid() 
plt.show()
```

# Polygon Clipping

```python
import matplotlib.pyplot as plt

xmin, xmax = 2, 8
ymin, ymax = 2, 8

polygon = [(1, 3), (5, 1), (9, 4), (7, 9), (3, 7)]

def inside(p, edge):
    x, y = p
    if edge == "left":
        return x >= xmin
    elif edge == "right":
        return x <= xmax
    elif edge == "bottom":
        return y >= ymin
    elif edge == "top":
        return y <= ymax

def intersect(p1, p2, edge):
    x1, y1 = p1
    x2, y2 = p2

    if edge == "left":
        x = xmin
        y = y1 + (y2 - y1) * (xmin - x1) / (x2 - x1)
    elif edge == "right":
        x = xmax
        y = y1 + (y2 - y1) * (xmax - x1) / (x2 - x1)
    elif edge == "bottom":
        y = ymin
        x = x1 + (x2 - x1) * (ymin - y1) / (y2 - y1)
    elif edge == "top":
        y = ymax
        x = x1 + (x2 - x1) * (ymax - y1) / (y2 - y1)

    return (x, y)

def sutherland_hodgman(polygon):
    edges = ["left", "right", "bottom", "top"]
    output = polygon

    for edge in edges:
        input_list = output
        output = []

        if not input_list:
            break

        prev = input_list[-1]

        for curr in input_list:
            if inside(curr, edge):
                if inside(prev, edge):
                    output.append(curr)
                else:
                    output.append(intersect(prev, curr, edge))
                    output.append(curr)
            elif inside(prev, edge):
                output.append(intersect(prev, curr, edge))

            prev = curr

    return output

clipped = sutherland_hodgman(polygon)

plt.figure()
plt.title("Polygon Clipping")

plt.plot([xmin, xmax, xmax, xmin, xmin],
         [ymin, ymin, ymax, ymax, ymin], 'k-')

x, y = zip(*(polygon + [polygon[0]]))
plt.plot(x, y, 'r--', label='Original Polygon')

if clipped:
    xc, yc = zip(*(clipped + [clipped[0]]))
    plt.plot(xc, yc, 'g-', label='Clipped Polygon')

plt.legend()
plt.grid()
plt.show()
```
