```python
import matplotlib.pyplot as plt

# Clipping window boundaries
xmin, xmax = 2, 8
ymin, ymax = 2, 8

# Sample points
points = [(1, 1), (3, 3), (5, 5), (9, 7), (6, 2)]

inside_points = []
outside_points = []

# Point clipping logic
for (x, y) in points:
    if xmin <= x <= xmax and ymin <= y <= ymax:
        inside_points.append((x, y))
    else:
        outside_points.append((x, y))

# Plotting
plt.figure()
plt.title("Point Clipping")

# Draw clipping window
plt.plot([xmin, xmax, xmax, xmin, xmin],
         [ymin, ymin, ymax, ymax, ymin], 'k-')

# Plot points
if inside_points:
    x_in, y_in = zip(*inside_points)
    plt.scatter(x_in, y_in, color='green', label='Inside')

if outside_points:
    x_out, y_out = zip(*outside_points)
    plt.scatter(x_out, y_out, color='red', label='Outside')

plt.legend()
plt.grid()
plt.show()
```
