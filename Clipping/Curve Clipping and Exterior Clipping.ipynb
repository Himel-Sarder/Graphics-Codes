```python
import numpy as np
import matplotlib.pyplot as plt

# Clipping window
xmin, xmax = -5, 5
ymin, ymax = -5, 5

# Generate curve (parabola)
x = np.linspace(-10, 10, 400)
y = x**2 / 5

# Separate inside and outside points
x_in, y_in = [], []
x_out, y_out = [], []

for xi, yi in zip(x, y):
    if xmin <= xi <= xmax and ymin <= yi <= ymax:
        x_in.append(xi)
        y_in.append(yi)
    else:
        x_out.append(xi)
        y_out.append(yi)

# Plot
plt.figure()
plt.title("Curve Clipping (Parabola)")

# Clipping window
plt.plot([xmin, xmax, xmax, xmin, xmin],
         [ymin, ymin, ymax, ymax, ymin], 'k-')

# Curve parts
plt.plot(x_out, y_out, 'r--', label='Outside Curve')
plt.plot(x_in, y_in, 'g-', label='Clipped Curve (Inside)')

plt.legend()
plt.grid()
plt.show()
```

```python
import numpy as np
import matplotlib.pyplot as plt

# Clipping window
xmin, xmax = -5, 5
ymin, ymax = -5, 5

# Generate curve (parabola)
x = np.linspace(-10, 10, 400)
y = x**2 / 4   # Parabola

# Separate exterior and interior points
x_ext, y_ext = [], []
x_int, y_int = [], []

for xi, yi in zip(x, y):
    if xi < xmin or xi > xmax or yi < ymin or yi > ymax:
        x_ext.append(xi)
        y_ext.append(yi)
    else:
        x_int.append(xi)
        y_int.append(yi)

# Plot
plt.figure()
plt.title("Exterior Clipping of Parabola")

# Clipping window
plt.plot([xmin, xmax, xmax, xmin, xmin],
         [ymin, ymin, ymax, ymax, ymin], 'k-')

# Exterior part (visible)
plt.plot(x_ext, y_ext, 'b-', label='Exterior Curve')

# Interior part (hidden/dashed)
plt.plot(x_int, y_int, 'r--', label='Interior (Hidden)')

plt.legend()
plt.grid()
plt.show()
```
