```python
import matplotlib.pyplot as plt

def bresenham_circle(xc, yc, r):
    x = 0
    y = r
    d = 3 - 2 * r   # initial decision parameter

    x_points = []
    y_points = []

    while x <= y:
        # 8-way symmetry points
        points = [
            ( x + xc,  y + yc),
            ( y + xc,  x + yc),
            (-x + xc,  y + yc),
            (-y + xc,  x + yc),
            (-x + xc, -y + yc),
            (-y + xc, -x + yc),
            ( x + xc, -y + yc),
            ( y + xc, -x + yc)
        ]

        # store points
        for px, py in points:
            x_points.append(px)
            y_points.append(py)

        if d < 0:
            d = d + 4 * x + 6
            x += 1
        else:
            d = d + 4 * (x - y) + 10
            x += 1
            y -= 1

    return x_points, y_points


# Input
xc = int(input("Enter center x: "))
yc = int(input("Enter center y: "))
r = int(input("Enter radius: "))

# Run algorithm
x, y = bresenham_circle(xc, yc, r)

# Print points
print("\nGenerated Points:")
for i in range(len(x)):
    print(f"({x[i]}, {y[i]})")

# Plot
plt.scatter(x, y)
plt.title("Bresenham Circle Drawing Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True)
plt.show()
```
