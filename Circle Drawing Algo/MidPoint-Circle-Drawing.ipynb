```python
import matplotlib.pyplot as plt

def midpoint_circle(xc, yc, r):
    x = 0
    y = r
    p = 1 - r

    x_points = []
    y_points = []

    while x <= y:
        # 8-way symmetry points
        points = [
            ( x + xc,  y + yc),
            ( y + xc,  x + yc),
            (-x + xc,  y + yc),
            (-y + xc,  x + yc),
            (-x + xc, -y + yc),
            (-y + xc, -x + yc),
            ( x + xc, -y + yc),
            ( y + xc, -x + yc)
        ]

        # store points
        for px, py in points:
            x_points.append(px)
            y_points.append(py)

        # Decision parameter update
        if p < 0:
            p = p + 2*x + 3
            x += 1
        else:
            p = p + 2*(x - y) + 5
            x += 1
            y -= 1

    return x_points, y_points

# Input
xc = int(input("Enter center x: "))
yc = int(input("Enter center y: "))
r = int(input("Enter radius: "))

# Run algorithm
x, y = midpoint_circle(xc, yc, r)

# Print generated points
print("\nGenerated Points:")
for i in range(len(x)):
    print(f"({x[i]}, {y[i]})")

# Plot
plt.scatter(x, y)
plt.title("Mid-Point Circle Drawing Algorithm")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.gca().set_aspect('equal', adjustable='box')
plt.grid(True)
plt.show()
```
