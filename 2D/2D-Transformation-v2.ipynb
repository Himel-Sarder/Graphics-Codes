# 2D-Translation

```python
import matplotlib.pyplot as plt

x = [2, 7, 4, 2]
y = [2, 2, 4, 2]

Tx = int(input('Enter Distance Tx : '))
Ty = int(input('Enter Distance Ty : '))

x_new = [xi + Tx for xi in x]
y_new = [yi + Ty for yi in y]

for x_p, y_p in zip(x_new, y_new):
    print(f'({x_p}, {y_p})')

plt.fill(x, y, 'b', alpha= 0.2, label='Original')
plt.plot(x, y, 'black')

plt.fill(x_new, y_new, 'r', alpha= 0.2, label='Translated')
plt.plot(x_new, y_new, 'black')
plt.legend()
plt.grid()
plt.show()
```

# 2D-Rotation

```python
import matplotlib.pyplot as plt
import math 

x = [2, 7, 4, 2]
y = [2, 2, 4, 2]

angle = int(input('Enter Rotation Angle : '))
radi = math.radians(angle)

cx = sum(x) / len(x)
cy = sum(y) / len(y)

x_new = [
    (xp - cx) * math.cos(radi) - (yp - cy) * math.sin(radi) + cx
    for xp, yp in zip(x, y)
]

y_new = [
    (xp - cx) * math.sin(radi) + (yp - cy) * math.cos(radi) + cy
    for xp, yp in zip(x, y)
]

for x_p, y_p in zip(x_new, y_new):
    print(f'({x_p}, {y_p})')

plt.fill(x, y, 'b', alpha= 0.2, label='Original')
plt.plot(x, y, 'black')

plt.fill(x_new, y_new, 'r', alpha= 0.2, label='Rotated')
plt.plot(x_new, y_new, 'black')
plt.legend()
plt.grid()
plt.show()
```

# 2D-Scaling

```python
import matplotlib.pyplot as plt

x = [2, 7, 4, 2]
y = [2, 2, 4, 2]

Sx = int(input('Enter Scaling Factor Sx : '))
Sy = int(input('Enter Scaling Factor Sy : '))

cx = sum(x) / len(x)
cy = sum(y) / len(y)

x_new = [(xi - cx) * Sx + cx for xi in x]
y_new = [(yi - cy) * Sy + cy for yi in y]

for x_p, y_p in zip(x_new, y_new):
    print(f'({x_p}, {y_p})')

plt.fill(x, y, 'b', alpha= 0.2, label='Original')
plt.plot(x, y, 'black')

plt.fill(x_new, y_new, 'r', alpha= 0.2, label='Scaled')
plt.plot(x_new, y_new, 'black')
plt.legend()
plt.grid()
plt.show()
```

# 2D-Reflection

```python
import matplotlib.pyplot as plt

# Original polygon
x = [2, 7, 4, 2]
y = [2, 2, 4, 2]

print("Choose Reflection Type:")
print("1. X-axis")
print("2. Y-axis")
print("3. Origin")

choice = int(input("Enter choice: "))

x_new = []
y_new = []

for xi, yi in zip(x, y):
    if choice == 1:  # X-axis
        x_new.append(xi)
        y_new.append(-yi)

    elif choice == 2:  # Y-axis
        x_new.append(-xi)
        y_new.append(yi)

    elif choice == 3:  # Origin
        x_new.append(-xi)
        y_new.append(-yi)

# Print reflected points
print("\nReflected Points:")
for xi, yi in zip(x_new, y_new):
    print(f"({xi}, {yi})")

# Plot
plt.fill(x, y, 'b', alpha=0.2, label='Original')
plt.fill(x_new, y_new, 'r', alpha=0.2, label='Reflected')

plt.plot(x, y, 'black')
plt.plot(x_new, y_new, 'black')

plt.title("2D Reflection")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.show()
```

# 2D-Shearing

```python
import matplotlib.pyplot as plt

# Original shape
x = [1, 7, 4, 1]
y = [0, 0, 5, 0]

print("Choose Shearing Type:")
print("1. X-direction Shearing")
print("2. Y-direction Shearing")
print("0. Exit")

choice = int(input("Enter choice: "))

if choice == 0:
    print("Program exited.")
    exit()

x_new = []
y_new = []

if choice == 1:
    shx = float(input("Enter shx (shear factor): "))
    for xi, yi in zip(x, y):
        x_new.append(xi + shx * yi)
        y_new.append(yi)

elif choice == 2:
    shy = float(input("Enter shy (shear factor): "))
    for xi, yi in zip(x, y):
        x_new.append(xi)
        y_new.append(yi + shy * xi)

else:
    print("Invalid choice!")
    exit()

# Print new points
print("\nSheared Points:")
for xi, yi in zip(x_new, y_new):
    print(f"({xi}, {yi})")

# Plot
plt.fill(x, y, alpha=0.5, label='Original')
plt.fill(x_new, y_new, alpha=0.5, label='Sheared')

plt.plot(x, y)
plt.plot(x_new, y_new)

plt.axhline(0)
plt.axvline(0)

plt.title("2D Shearing")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')

plt.show()
```
