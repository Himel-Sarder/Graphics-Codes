# 2D-Translation

```python
import matplotlib.pyplot as plt

# Original points (example: triangle)
x = [2, 4, 3, 2]  # last point same as first for closed polygon
y = [3, 3, 5, 3]

# Translation values
tx = int(input("Enter translation in X (tx): "))
ty = int(input("Enter translation in Y (ty): "))

# Apply translation
x_new = [xi + tx for xi in x]
y_new = [yi + ty for yi in y]

# Print new points
print("\nTranslated Points:")
for xi, yi in zip(x_new, y_new):  # (4, 4), (7, 4), (5, 7), (4, 4)
    print(f"({xi}, {yi})")

# Plot original and translated shapes
plt.fill(x, y, alpha=0.5, label='Original')     # original filled
plt.fill(x_new, y_new, alpha=0.5, label='Translated')
plt.plot(x, y)
plt.plot(x_new, y_new)
plt.title("2D Translation")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.show()
```

# 2D-Rotation

```python
import matplotlib.pyplot as plt 
import math

x = [1, 4, 2, 1]  # closed polygon
y = [2, 2, 5, 2]

angle_deg = float(input("Enter rotation angle in degrees: "))
angle_rad = math.radians(angle_deg)  # convert to radians

# Rotation formula
xnew = [xi * math.cos(angle_rad) - yi * math.sin(angle_rad) for xi, yi in zip(x, y)]
ynew = [xi * math.sin(angle_rad) + yi * math.cos(angle_rad) for xi, yi in zip(x, y)]

# Print rotated points
print('\nRotated Points : \n')
for xi, yi in zip(xnew, ynew):
    print(f'({round(xi,2)}, {round(yi,2)})')

# Plot original and rotated polygons
plt.fill(x, y, alpha=0.5, label='Original')     # original filled
plt.fill(xnew, ynew, alpha=0.5, label='Rotated')  # rotated filled
plt.plot(x, y)
plt.plot(xnew, ynew)
plt.title("2D Rotation")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.show()
```

# 2D Scaling

```python
import matplotlib.pyplot as plt

# Original points (example: triangle)
x = [2, 4, 3, 2]  # last point same as first for closed polygon
y = [3, 3, 5, 3]

# Translation values
sx = int(input("Enter Scaling factor for X-axis -> Sx: "))
sy = int(input("Enter Scaling factor for y-axis -> Sy: "))

# Apply scaling
x_new = [xi * sx for xi in x]
y_new = [yi * sy for yi in y]

# Print new points
print("\nScaled Points:")
for xi, yi in zip(x_new, y_new):  # (4, 4), (7, 4), (5, 7), (4, 4)
    print(f"({xi}, {yi})")

# Plot original and scaled shapes
plt.fill(x, y, alpha=0.5, label='Original')     # original filled
plt.fill(x_new, y_new, alpha=0.5, label='Scaled')
plt.plot(x, y)
plt.plot(x_new, y_new)
plt.title("2D Scaling")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.show()
```

# 2D Reflection

```python
import matplotlib.pyplot as plt

# Original polygon
x = [1, 7, 4, 1]
y = [0, 0, 5, 0]

print("Choose Reflection Type:")
print("1. X-axis")
print("2. Y-axis")
print("3. Origin")
print("4. Line y = x")
print("5. Line y = -x")

choice = int(input("Enter choice: "))

x_new = []
y_new = []

for xi, yi in zip(x, y):

    if choice == 1:  # X-axis
        x_new.append(xi)
        y_new.append(-yi)

    elif choice == 2:  # Y-axis
        x_new.append(-xi)
        y_new.append(yi)

    elif choice == 3:  # Origin
        x_new.append(-xi)
        y_new.append(-yi)

    elif choice == 4:  # y = x
        x_new.append(yi)
        y_new.append(xi)

    elif choice == 5:  # y = -x
        x_new.append(-yi)
        y_new.append(-xi)

# Print reflected points
print("\nReflected Points:")
for xi, yi in zip(x_new, y_new):
    print(f"({xi}, {yi})")

# Plot
plt.fill(x, y, alpha=0.5, label='Original')
plt.fill(x_new, y_new, alpha=0.5, label='Reflected')

plt.plot(x, y)
plt.plot(x_new, y_new)

plt.axhline(0)
plt.axvline(0)

plt.title("2D Reflection")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')
plt.show()
```

# 2D Shearing

```python
import matplotlib.pyplot as plt

# Original shape
x = [1, 7, 4, 1]
y = [0, 0, 5, 0]

print("Choose Shearing Type:")
print("1. X-direction Shearing")
print("2. Y-direction Shearing")
print("0. Exit")

choice = int(input("Enter choice: "))

if choice == 0:
    print("Program exited.")
    exit()

x_new = []
y_new = []

if choice == 1:
    shx = float(input("Enter shx (shear factor): "))
    for xi, yi in zip(x, y):
        x_new.append(xi + shx * yi)
        y_new.append(yi)

elif choice == 2:
    shy = float(input("Enter shy (shear factor): "))
    for xi, yi in zip(x, y):
        x_new.append(xi)
        y_new.append(yi + shy * xi)

else:
    print("Invalid choice!")
    exit()

# Print new points
print("\nSheared Points:")
for xi, yi in zip(x_new, y_new):
    print(f"({xi}, {yi})")

# Plot
plt.fill(x, y, alpha=0.5, label='Original')
plt.fill(x_new, y_new, alpha=0.5, label='Sheared')

plt.plot(x, y)
plt.plot(x_new, y_new)

plt.axhline(0)
plt.axvline(0)

plt.title("2D Shearing")
plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.grid(True)
plt.legend()
plt.gca().set_aspect('equal', adjustable='box')

plt.show()
```
